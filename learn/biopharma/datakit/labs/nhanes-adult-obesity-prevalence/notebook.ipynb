# Survey-weighted obesity prevalence from NHANES

**Data kit lab** `nhanes-adult-obesity-prevalence` · Real-world · about 45 minutes

**The question.** What share of US adults age 20 and older had obesity (BMI 30 or more) and severe obesity (BMI 40 or more) in August 2021 to August 2023, overall, by sex and by age group?

**The data.** Two public-use files of the National Health and Nutrition Examination Survey (NHANES), cycle August 2021 to August 2023, published by the National Center for Health Statistics (NCHS): demographics with the survey weights (DEMO_L) and body measures (BMX_L). The data are real survey records with identifiers removed. The NCHS data user agreement allows statistical reporting and analysis only: do not try to identify anyone, and do not link the files to identifiable data.

**The published result.** NCHS Data Brief no. 508 (September 2024), data tables for Figure 1 (obesity, page 7) and Figure 3 (severe obesity, page 8).

**What you do.**

1. Read the two files and join them on the respondent number.
2. Mark the adults the brief analyses (examined, age 20 and older, not pregnant, BMI measured).
3. Compute weighted percents with the exam weight.
4. Compute standard errors that respect the survey design (strata and PSUs).
5. Age-adjust to the Census 2000 standard population.
6. Compare all 72 numbers with the brief.

## 1. Read and join the files

NHANES ships one SAS transport file per survey component. Every file carries `SEQN`, the respondent sequence number. The demographics file has one row per sampled person; the body measures file has one row per examined person.

```python
import numpy as np
import pandas as pd

demo = pd.read_sas(data_file("demo"), format="xport")
bmx = pd.read_sas(data_file("bmx"), format="xport")
print("DEMO_L:", demo.shape, " BMX_L:", bmx.shape)

df = demo.merge(bmx[["SEQN", "BMXBMI"]], on="SEQN", how="left", validate="one_to_one")
print("joined:", df.shape)
df[["SEQN", "RIDSTATR", "RIAGENDR", "RIDAGEYR", "RIDEXPRG", "WTMEC2YR", "SDMVSTRA", "SDMVPSU", "BMXBMI"]].head()
```

## 2. Who is in the analysis?

The brief's methods section: all examined participants age 20 and older; pregnant women and adults with missing height or weight excluded.

- `RIDSTATR == 2`: interviewed and examined at the mobile examination centre.
- `RIDAGEYR >= 20`.
- `RIDEXPRG == 1` is pregnant. Code 2 is not pregnant and code 3 is "could not be determined". Only code 1 is excluded.
- `BMXBMI` present. BMI is weight in kg over height in metres squared, rounded to one decimal (the brief's definition).

We keep every row of the file and use a 0/1 flag for the analysis group. The reason comes in section 4.

```python
print(df["RIDSTATR"].value_counts().sort_index().rename("RIDSTATR"))
print(df["RIDEXPRG"].value_counts(dropna=False).sort_index().rename("RIDEXPRG"))

df["bmi"] = df["BMXBMI"].round(1)
changed = int(((df["BMXBMI"] >= 30) != (df["bmi"] >= 30)).sum() + ((df["BMXBMI"] >= 40) != (df["bmi"] >= 40)).sum())
print("rows whose class changes when BMI is rounded to one decimal first:", changed)

examined_adult = (df["RIDSTATR"] == 2) & (df["RIDAGEYR"] >= 20)
pregnant = df["RIDEXPRG"] == 1
df["in_analysis"] = examined_adult & ~pregnant & df["bmi"].notna()
print("examined adults 20+:", int(examined_adult.sum()))
print("  of whom pregnant:", int((examined_adult & pregnant).sum()))
print("  of whom BMI missing (not pregnant):", int((examined_adult & ~pregnant & df["bmi"].isna()).sum()))
print("analysis group:", int(df["in_analysis"].sum()))
```

## 3. Weighted percents

Each examined person carries an exam weight `WTMEC2YR`: the number of people in the US population that person stands for. The codebook gives weight 0 to people who were interviewed but not examined. Look at what pandas reads for them below: these zeros arrive as a tiny positive number. It does no harm here because they are outside the analysis group, but never test a weight with `== 0`. BMI is measured at the exam, so the exam weight is the right one; the interview weight `WTINT2YR` is not.

The weighted percent in a group is the sum of weights of people with obesity divided by the sum of weights of everyone in the group.

```python
w = df["WTMEC2YR"].to_numpy(float)
print("weight range among examined:", w[df["RIDSTATR"] == 2].min().round(1), "to", w[df["RIDSTATR"] == 2].max().round(1))
print("weight of the not examined, as read:", np.unique(w[df["RIDSTATR"] == 1]))
print("how many of them test == 0:", int((w[df["RIDSTATR"] == 1] == 0).sum()))

obese = (df["bmi"] >= 30).astype(float).to_numpy()
severe = (df["bmi"] >= 40).astype(float).to_numpy()
d = df["in_analysis"].to_numpy(float)

unweighted = obese[d == 1].mean()
weighted = (w * d * obese).sum() / (w * d).sum()
print(f"obesity, unweighted share of the sample: {100 * unweighted:.1f}%")
print(f"obesity, weighted estimate for US adults: {100 * weighted:.1f}%")
```

## 4. Standard errors from the survey design

NHANES is a multistage sample: counties first, then areas, households and people. People in the same area resemble each other, so the sample carries less information than a simple random sample of the same size. The public file gives masked design variables for this: stratum `SDMVSTRA` and primary sampling unit (PSU) `SDMVPSU`.

The brief names SUDAAN and the R survey package. Both use Taylor-series linearisation by default. We write it out:

1. For a weighted percent `p` in a group with indicator `d`, each person's linearised value is `z = w * d * (y - p) / sum(w * d)`.
2. Sum `z` within each PSU.
3. In each stratum with `n` PSUs, add `n / (n - 1)` times the sum of squared deviations of the PSU totals from their mean.
4. The standard error is the square root of the total.

A subgroup (men 40-59, say) is a *domain*. `z` is 0 for everyone outside it, but every PSU stays in the sum. If you cut the file down to the subgroup first, a PSU with no members drops out and the standard error can come out wrong.

```python
strata = df["SDMVSTRA"].to_numpy()
psu = df["SDMVPSU"].to_numpy()
print("strata:", int(strata.min()), "to", int(strata.max()), "| PSUs per stratum:", df.groupby("SDMVSTRA")["SDMVPSU"].nunique().unique())


def domain_percent(y, d):
    """Weighted mean of y in the domain d, and each row's linearised value."""
    wd = w * d
    p = (wd * y).sum() / wd.sum()
    return p, wd * (y - p) / wd.sum()


def taylor_se(z):
    """With-replacement Taylor-series standard error from linearised values z."""
    totals = pd.DataFrame({"h": strata, "psu": psu, "z": z}).groupby(["h", "psu"])["z"].sum()
    var = 0.0
    for _, g in totals.groupby(level="h"):
        n = len(g)
        var += n / (n - 1) * ((g - g.mean()) ** 2).sum()
    return float(np.sqrt(var))


p, z = domain_percent(obese, d)
print(f"obesity, all adults 20+: {100 * p:.2f}%  SE {100 * taylor_se(z):.2f}")
srs = np.sqrt(p * (1 - p) / d.sum())
print(f"SE if this were a simple random sample of the same size: {100 * srs:.2f}")
```

## 5. All groups, and age adjustment

The brief reports each sex by age group (20-39, 40-59, 60 and older) and for 20 and older. For 20 and older it also gives an age-adjusted percent: the age-specific percents weighted by the share of each age group in the US Census 2000 population. This removes differences in age mix, so that men and women, or two survey cycles, can be compared.

The standard counts (thousands) come from the NHANES sample code tutorial: 77,670 aged 20-39, 72,816 aged 40-59 and 45,364 aged 60 and older, out of 195,850. The adjusted percent's linearised value is the same weighted sum of the age groups' linearised values.

```python
age = df["RIDAGEYR"]
AGE_GROUPS = {"20plus": age >= 20, "20_39": (age >= 20) & (age < 40), "40_59": (age >= 40) & (age < 60), "60plus": age >= 60}
SEXES = {"total": pd.Series(True, index=df.index), "men": df["RIAGENDR"] == 1, "women": df["RIAGENDR"] == 2}
CENSUS_2000 = {"20_39": 77670, "40_59": 72816, "60plus": 45364}
STANDARD = {k: v / sum(CENSUS_2000.values()) for k, v in CENSUS_2000.items()}
print("standard proportions:", {k: round(v, 6) for k, v in STANDARD.items()})

RESULTS, rows = {}, []
for tag, y in (("obesity", obese), ("severe", severe)):
    for sex, in_sex in SEXES.items():
        parts = {}
        for ag, in_age in AGE_GROUPS.items():
            dom = (df["in_analysis"] & in_sex & in_age).to_numpy(float)
            p, z = domain_percent(y, dom)
            parts[ag] = (p, z)
            RESULTS[f"{tag}_pct_{sex}_{ag}"] = 100 * p
            RESULTS[f"{tag}_se_{sex}_{ag}"] = 100 * taylor_se(z)
            RESULTS[f"n_{sex}_{ag}"] = int(dom.sum())
            rows.append((tag, sex, ag, int(dom.sum()), 100 * p, 100 * taylor_se(z)))
        p_adj = sum(STANDARD[ag] * parts[ag][0] for ag in STANDARD)
        z_adj = sum(STANDARD[ag] * parts[ag][1] for ag in STANDARD)
        RESULTS[f"{tag}_aa_pct_{sex}"] = 100 * p_adj
        RESULTS[f"{tag}_aa_se_{sex}"] = 100 * taylor_se(z_adj)

table = pd.DataFrame(rows, columns=["outcome", "sex", "age", "n", "percent", "se"])
print(table.round(1).to_string(index=False))
print({k: round(v, 1) for k, v in RESULTS.items() if "_aa_" in k})
```
