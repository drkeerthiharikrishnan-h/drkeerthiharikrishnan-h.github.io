## 1. Load the five tables

Ids are 64-bit integers, some negative. Read them as text so that nothing is rounded. Dates are written as `YYYY-MM-DD`; they run into the 2100s because de-identification shifts every patient's dates.

```python
import sqlite3

import pandas as pd

TABLES = ["person", "observation_period", "visit_occurrence", "drug_exposure", "death"]
frames = {t: pd.read_csv(data_file(t.replace("_", "-")), dtype=str, keep_default_na=False) for t in TABLES}
for t, df in frames.items():
    print(f"{t:<20}{len(df):>8} rows  {len(df.columns):>3} columns")
print()
print("visit dates from", frames["visit_occurrence"]["visit_start_date"].min(), "to", frames["visit_occurrence"]["visit_start_date"].max())
```

## 2. Put them in a SQL database

SQLite comes with Python. Concept ids become integers; dates stay as ISO text, so comparing text compares dates.

```python
con = sqlite3.connect(":memory:")
INT_COLS = {"gender_concept_id", "visit_concept_id", "drug_concept_id", "death_type_concept_id"}
for t, df in frames.items():
    df = df.copy()
    for c in INT_COLS & set(df.columns):
        df[c] = df[c].astype(int)
    df.to_sql(t, con, index=False)
q = lambda sql: pd.read_sql_query(sql, con)
print(q("SELECT name FROM sqlite_master WHERE type='table'"))
```

## 3. Persons and gender (Achilles analyses 1 and 2)

Analysis 1 counts distinct `person_id` in PERSON. Analysis 2 counts them by `gender_concept_id`; the export keeps 8507 (MALE) and 8532 (FEMALE). Neither joins the observation period.

```python
RESULTS = {}
RESULTS["n_persons"] = int(q("SELECT COUNT(DISTINCT person_id) AS n FROM person").n[0])
g = q("SELECT gender_concept_id, COUNT(DISTINCT person_id) AS n FROM person GROUP BY gender_concept_id")
print(g)
gd = dict(zip(g.gender_concept_id, g.n))
RESULTS["n_female"], RESULTS["n_male"] = int(gd.get(8532, 0)), int(gd.get(8507, 0))
```

## 4. The observation-period join

In the Achilles code that was run (commit c1136949, version 1.6.7), visits, drug exposures and deaths count only if their start date lies inside one of the same person's observation periods. One SQL fragment does that for every event table.

```python
def in_period(table, date_col):
    return f"""FROM {table} e JOIN observation_period op ON e.person_id = op.person_id
              AND e.{date_col} >= op.observation_period_start_date
              AND e.{date_col} <= op.observation_period_end_date"""

print(q("SELECT COUNT(*) AS periods, COUNT(DISTINCT person_id) AS persons FROM observation_period"))
print(q("SELECT person_id, COUNT(*) AS n FROM observation_period GROUP BY person_id ORDER BY n DESC LIMIT 3"))
```

## 5. Deaths by type (analysis 505)

`COUNT(person_id)` counts death records, not distinct persons.

```python
d = q(f"SELECT e.death_type_concept_id AS c, COUNT(e.person_id) AS n {in_period('death', 'death_date')} GROUP BY e.death_type_concept_id")
print(d)
RESULTS["n_death_records_ehr"] = int(dict(zip(d.c, d.n)).get(32817, 0))
print("death rows in the table:", len(frames["death"]))
```

## 6. Visits by visit concept (analyses 200 and 201)

200 counts distinct persons with a visit of each concept; 201 counts visit records. Records per person is 201 divided by 200, not rounded.

```python
VISIT_NAMES = {k.split("_")[-1]: c["label"] for k, c in PUBLISHED.items() if k.startswith("visit_persons_")}
v = q(f"""SELECT e.visit_concept_id AS c, COUNT(DISTINCT e.person_id) AS persons, COUNT(e.person_id) AS records
          {in_period('visit_occurrence', 'visit_start_date')} GROUP BY e.visit_concept_id ORDER BY persons DESC""")
v["records_per_person"] = v.records / v.persons
print(v.to_string(index=False))
for r in v.itertuples(index=False):
    if str(int(r.c)) in VISIT_NAMES:
        RESULTS[f"visit_persons_{int(r.c)}"] = int(r.persons)
        RESULTS[f"visit_rpp_{int(r.c)}"] = float(r.records) / float(r.persons)
```

## 7. Drug exposures by drug concept (analyses 700 and 701)

The same two counts on DRUG_EXPOSURE, by `drug_concept_id`. The drug export rounds records per person to five decimals. The checks are the three drug concepts with the most persons in the published file.

```python
de = q(f"""SELECT e.drug_concept_id AS c, COUNT(DISTINCT e.person_id) AS persons, COUNT(e.person_id) AS records
           {in_period('drug_exposure', 'drug_exposure_start_date')} GROUP BY e.drug_concept_id ORDER BY persons DESC""")
de["records_per_person"] = (de.records / de.persons).round(5)
print(len(de), "drug concepts with at least one exposure inside a period")
print(de.head(6).to_string(index=False))
DRUGS = [k.split("_")[-1] for k in PUBLISHED if k.startswith("drug_persons_")]
for r in de.itertuples(index=False):
    if str(int(r.c)) in DRUGS:
        RESULTS[f"drug_persons_{int(r.c)}"] = int(r.persons)
        RESULTS[f"drug_rpp_{int(r.c)}"] = round(float(r.records) / float(r.persons), 5)
```

## 8. Shown, not checked: what the definitions change

The same counts without the observation-period join, and the cells Achilles would delete (counts of 5 or fewer).

```python
for table, date_col, ccol in (("visit_occurrence", "visit_start_date", "visit_concept_id"),
                              ("drug_exposure", "drug_exposure_start_date", "drug_concept_id"),
                              ("death", "death_date", "death_type_concept_id")):
    all_rows = int(q(f"SELECT COUNT(*) AS n FROM {table}").n[0])
    kept = int(q(f"SELECT COUNT(*) AS n {in_period(table, date_col)}").n[0])
    print(f"{table:<18} rows {all_rows:>7}, inside an observation period {kept:>7}, outside {all_rows - kept:>6}")
print()
print("drug concepts with 5 or fewer persons (deleted from the Achilles results):", int((de.persons <= 5).sum()), "of", len(de))
print("race concepts in PERSON:")
print(q("SELECT race_concept_id, COUNT(DISTINCT person_id) AS n FROM person GROUP BY race_concept_id ORDER BY n DESC"))
```

## 9. Compare with the published values

```python
report(RESULTS)
```

## What to take away

- A count in a common data model needs a definition: distinct persons or records, and which time window.
- Achilles counts events only inside observation periods and deletes small cells. Rebuild a published characterization with the same rules, or the numbers will not line up.
- The two parts of this release carry different dates (Achilles 2021-04-26, CSVs 2021-05-17). Check dates before you read a match or a mismatch.
