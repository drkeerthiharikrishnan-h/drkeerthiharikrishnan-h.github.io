# Marker genes of two liver macrophage populations, from a single-cell atlas

**Data kit lab** `liver-macrophage-markers` · Discovery · about 40 minutes

**The question.** Which genes separate the two CD68-positive macrophage populations of the human liver, the inflammatory one and the non-inflammatory one?

**What you do.**

1. Open a curated single-cell file (AnnData, h5ad) with `h5py` and look at its parts.
2. Find the two macrophage populations by the authors' labels.
3. Normalise the raw counts per cell.
4. For each gene the paper lists, test whether it is higher in its population than in all other cells.
5. Compare the two populations with each other.

**The data.** Source: CZ CELLxGENE Discover, dataset "Liver" of the collection by MacParland and others, under CC BY 4.0 (attribute the contributors and CELLxGENE). Nothing is copied into this notebook: the first cell downloads the file from the source. **The file is about 65 MB**, the largest download in this kit; allow a minute or more.

## 1. What is inside an h5ad file

An h5ad file is an HDF5 file with fixed groups: `obs` (one row per cell), `var` (one row per gene), `X` (the matrix), and often `raw` (the raw counts with their own `var`). `h5py` opens it without loading the matrix into memory.

```python
import h5py
import numpy as np
import pandas as pd
import scipy.sparse as sp

h5 = h5py.File(data_file("h5ad"), "r")
print("groups:", list(h5.keys()))
print("obs columns:", list(h5["obs"].keys()))
print("var columns:", list(h5["var"].keys()))
for name in ("X", "raw/X"):
    print(name, dict(h5[name].attrs)["encoding-type"], tuple(h5[name].attrs["shape"]))
```

A column in `obs` or `var` is either a plain array or a categorical (integer `codes` pointing into `categories`). A small helper reads both.

```python
def column(group, name):
    item = group[name]
    if isinstance(item, h5py.Group):  # categorical
        cats = np.array([c.decode() if isinstance(c, bytes) else c for c in item["categories"][()]], dtype=object)
        codes = item["codes"][()]
        return pd.Series(np.where(codes >= 0, cats[np.clip(codes, 0, None)], None))
    return pd.Series([v.decode() if isinstance(v, bytes) else v for v in item[()]])

cell_label = column(h5["obs"], "CellType")
print(len(cell_label), "cells")
labels = pd.DataFrame({"CellType (authors)": cell_label,
                       "BroadCellType (authors)": column(h5["obs"], "BroadCellType"),
                       "cell_type (Cell Ontology)": column(h5["obs"], "cell_type")})
labels.value_counts().rename("cells").reset_index()
```

The authors' label `CellType` names the two populations "Inflammatory Macs" and "Non-inflammatory Macs". The ontology column calls them "M1 macrophage" and "macrophage", and the broad label puts both under "Kupffer". Use the authors' label. The paper calls them CD68+ population 1 (Cluster 4) and population 2 (Cluster 10); the file has no cluster numbers.

```python
POPULATIONS = {"infl": "Inflammatory Macs", "noninfl": "Non-inflammatory Macs"}
in_pop = {}
for key, label in POPULATIONS.items():
    in_pop[key] = (cell_label == label).to_numpy()
    assert in_pop[key].sum() > 0, label
    print(f"{label}: {in_pop[key].sum()} cells; rest: {(~in_pop[key]).sum()} cells")
```

## 2. Raw counts, then normalise

`X` holds the authors' normalised values; `raw/X` holds the counts. Read `raw/X` as a sparse matrix and check that it holds whole numbers. Then normalise each cell to 10,000 counts and take log(1 + x). (The paper used scran's pooling method; this lab uses the simpler library-size method.)

```python
def csr(group):
    return sp.csr_matrix((group["data"][()], group["indices"][()], group["indptr"][()]), shape=tuple(group.attrs["shape"]))

counts = csr(h5["raw/X"])
assert np.all(counts.data == np.round(counts.data)), "raw/X is not whole counts"
x_sample = h5["X"]["data"][:1000]
print("raw/X whole numbers: yes; first values of X:", np.round(x_sample[:5], 3))
assert column(h5["var"], "_index").equals(column(h5["raw/var"], "_index"))
library = np.asarray(counts.sum(axis=1)).ravel()
print("counts per cell: median", int(np.median(library)), "min", int(library.min()))
```

## 3. Find the printed genes

The paper used the Cell Ranger GRCh38 1.2.0 gene names. The file has two name columns: `gene_symbols` (from the authors) and `feature_name` (from CELLxGENE). Look a name up in the authors' column first, case-insensitively, and only then in `feature_name`. Report a name that is not found instead of dropping it.

```python
author_names = column(h5["var"], "gene_symbols").str.upper()
cxg_names = column(h5["var"], "feature_name").str.upper()

def find_gene(name):
    for col_name, names in (("gene_symbols", author_names), ("feature_name", cxg_names)):
        hits = np.flatnonzero(names.to_numpy() == name.upper())
        if len(hits) == 1:
            return int(hits[0]), col_name
        if len(hits) > 1:
            return None, "ambiguous"
    return None, "absent"

# PUBLISHED carries each check's label, which starts with the gene name exactly as printed
printed = {key: [(k, v["label"].split(" higher in ")[0]) for k, v in PUBLISHED.items() if k.startswith(key + "_")] for key in POPULATIONS}
lookup = {name: find_gene(name) for key in printed for _, name in printed[key]}
pd.Series({name: where for name, (_, where) in lookup.items()}).value_counts()
```

## 4. Higher in the population than in all other cells?

The authors' code (cited in the paper's Methods) finds each cluster's DE genes by comparing the cluster with all other cells and keeping genes that are higher. Do the same for the direction: mean log-normalised expression in the population minus the mean in all other cells.

```python
def log_norm(columns):
    sub = counts[:, columns].toarray()
    return np.log1p(sub / library[:, None] * 1e4)

rows, RESULTS = [], {}
for key, genes in printed.items():
    mask = in_pop[key]
    for check, name in genes:
        idx, where = lookup[name]
        if idx is None:
            RESULTS[check] = where
            rows.append({"population": POPULATIONS[key], "gene": name, "found_in": where})
            continue
        e = log_norm([idx]).ravel()
        diff = e[mask].mean() - e[~mask].mean()
        RESULTS[check] = "higher" if diff > 0 else "not higher"
        rows.append({"population": POPULATIONS[key], "gene": name, "found_in": where,
                     "mean_pop": e[mask].mean(), "mean_rest": e[~mask].mean(), "difference": diff,
                     "detected_pop": (e[mask] > 0).mean(), "detected_rest": (e[~mask] > 0).mean(), "result": RESULTS[check]})
table = pd.DataFrame(rows)
table.round(3)
```

## 5. Inflammatory against non-inflammatory, directly

The question asks which genes *separate* the two populations. The printed lists answer a slightly different question (each population against all other cells). Compare the two populations with each other for the same genes. This is shown, not checked.

```python
both = sorted({name for genes in printed.values() for _, name in genes if lookup[name][0] is not None})
idx = [lookup[n][0] for n in both]
e = log_norm(idx)
direct = pd.DataFrame({"gene": both,
                       "listed_for": ["inflammatory" if any(n == g for _, g in printed["infl"]) else "non-inflammatory" for n in both],
                       "mean_infl": e[in_pop["infl"]].mean(axis=0), "mean_noninfl": e[in_pop["noninfl"]].mean(axis=0)})
direct["infl_minus_noninfl"] = direct["mean_infl"] - direct["mean_noninfl"]
direct["agrees_with_list"] = np.where(direct["listed_for"] == "inflammatory", direct["infl_minus_noninfl"] > 0, direct["infl_minus_noninfl"] < 0)
print(direct.groupby("listed_for")["agrees_with_list"].agg(["sum", "count"]))
direct.sort_values("infl_minus_noninfl").round(3)
```

## 6. Check against the paper

`PUBLISHED` holds one check per printed gene, copied by code from the Results text, each with the value "higher", plus the cell count from the abstract. `report` prints your result beside each one.

```python
RESULTS["n_cells"] = int(len(cell_label))
h5.close()
report(RESULTS)
```
