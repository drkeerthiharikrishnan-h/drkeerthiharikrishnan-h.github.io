**What goes wrong:** the first rows of a PubChem data table are not data. They give the type, description and unit of each column. Their tag starts with `RESULT_`. Count them and set them aside.

```python
is_meta = table["PUBCHEM_RESULT_TAG"].str.startswith("RESULT_")
meta = table[is_meta].set_index("PUBCHEM_RESULT_TAG")
rows = table[~is_meta]
print("rows that describe columns:", list(meta.index))
print("substance rows:", len(rows), "| different substance ids (SID):", rows["PUBCHEM_SID"].nunique())
print("activity call per substance:", rows["PUBCHEM_ACTIVITY_OUTCOME"].value_counts().to_dict())
```

## 3. Look at how the columns are organised

A substance can be tested more than once. Each test is a *replicate* and has its own block of columns, marked `-Replicate_1`, `-Replicate_2` and so on. A block holds the depositor's summary and fit, followed by one column per concentration. Block 1 has one extra field, `Compound QC`.

```python
replicates = sorted({int(c.rsplit("_", 1)[1]) for c in table.columns if "-Replicate_" in c})
block1 = [c for c in table.columns if c.endswith("-Replicate_1")]
fit_fields = [c.replace("-Replicate_1", "") for c in block1 if not c.startswith("Activity at")]
block_sizes = {k: sum(c.endswith(f"-Replicate_{k}") for c in table.columns) for k in replicates}
print("replicate blocks:", len(replicates), "| columns in block 1:", block_sizes[1], "| in every other block:", sorted({v for k, v in block_sizes.items() if k != 1}))
print("summary and fit fields:", fit_fields)
for name in ["Fit_LogAC50", "Excluded_Points", "Curve_Description"]:
    print("-", name + ":", meta.loc["RESULT_DESCR", name + "-Replicate_1"])
```

## 4. Get the concentrations and put them on the right scale

The concentrations are written in the column names, in micromolar (uM). The depositor's `Fit_LogAC50` is the log of the AC50 in molar units.

**What goes wrong:** if you compare a log of micromolar with a log of molar, every value is off by exactly 6. We convert once, here.

```python
conc_text = [re.fullmatch(r"Activity at ([\d.]+) uM-Replicate_1", c).group(1) for c in block1 if c.startswith("Activity at")]
conc_um = np.array([float(c) for c in conc_text])
log_conc = np.log10(conc_um * 1e-6)          # uM -> mol/L -> log10
print("concentration columns per block:", len(conc_um))
print("lowest and highest (uM):", conc_um.min(), conc_um.max())
print("the same as log10 mol/L:", log_conc.min().round(3), log_conc.max().round(3))
for k in replicates:                          # every block must use the same concentration columns
    assert all(f"Activity at {c} uM-Replicate_{k}" in table.columns for c in conc_text)
```

## 5. Reshape: one row per curve

A curve is one substance in one replicate. We stack the 46 blocks on top of each other and keep the blocks that hold data.

```python
blocks = []
for k in replicates:
    suffix = f"-Replicate_{k}"
    cols = [c for c in rows.columns if c.endswith(suffix)]
    block = rows[["PUBCHEM_SID", "PUBCHEM_ACTIVITY_OUTCOME"] + cols].copy()
    block.columns = ["sid", "substance_outcome"] + [c[: -len(suffix)] for c in cols]
    block["replicate"] = k
    blocks.append(block[block["Fit_CurveClass"].notna()])
curves = pd.concat(blocks, ignore_index=True)

activity_cols = [f"Activity at {c} uM" for c in conc_text]
Y = curves[activity_cols].apply(pd.to_numeric).to_numpy()     # response in percent, one row per curve
tested = ~np.isnan(Y)                                         # which of the 25 concentrations were used
print("curves:", len(curves))
print("tested concentrations per curve:", sorted(set(tested.sum(axis=1))))
print("curves per replicate block, first five blocks:", curves["replicate"].value_counts().sort_index().head(5).to_dict())
gaps = np.concatenate([np.diff(log_conc[t]) for t in tested])   # log10 gap between neighbouring tested concentrations
print("gap between neighbouring concentrations of a curve (log10): typical", np.median(gaps).round(3), "| smallest", gaps.min().round(3), "| largest", gaps.max().round(3))
```

Every curve has 15 points, but the table has 25 concentration columns. Different substances were tested over different ranges, so each curve fills 15 of the 25 columns and leaves the rest empty. The typical gap is one dilution step. The few much smaller and larger gaps sit at the top of the range, where the column labels are unevenly spaced.

## 6. Choose the curves to refit

The depositor gave each curve a class. The record explains the words: a *complete* curve has two observed plateaus (asymptotes); a *partial* curve may not have reached its second plateau.

This matters for the AC50. The AC50 is the concentration that gives half of the maximal response. If both plateaus were measured, the AC50 lies between tested concentrations and the points should fix it. Section 14 tests whether they do. If the top plateau was never reached, the AC50 depends on how a program guesses the top.

So this lab refits one class only: **1.1, "Complete curve; high efficacy"**.

**What goes wrong:** the class is stored as text, and the same class is written in two ways.

```python
print("ways class 1.1 is written:", curves.loc[curves["Fit_CurveClass"].str.startswith("1.1"), "Fit_CurveClass"].value_counts().to_dict())
curves["curve_class"] = pd.to_numeric(curves["Fit_CurveClass"])           # compare as a number, not as text
for col in ["Fit_LogAC50", "Fit_HillSlope", "Fit_ZeroActivity", "Fit_InfiniteActivity", "Fit_R2"]:
    curves[col] = pd.to_numeric(curves[col])

print(curves.groupby("curve_class")["Curve_Description"].agg(lambda s: s.dropna().unique().tolist() or ["(inactive, no fit)"]).to_string())
```

```python
scope = curves[curves["curve_class"] == 1.1]
print("curves in scope:", len(scope), "| substances:", scope["sid"].nunique())
print("their phenotype:", scope["Phenotype"].value_counts().to_dict())
print("overall call of the substance they belong to:", scope["substance_outcome"].value_counts().to_dict())
assert len(scope) == len(PUBLISHED), "the file changed: the checks no longer match the curves"
```

**What goes wrong:** the activity call (`PUBCHEM_ACTIVITY_OUTCOME`) is one call for the whole substance. It is not a property of a single curve. Some good-looking curves belong to substances called Inactive or Inconclusive overall. Select curves by their own class.

## 7. Points the depositor left out

`Excluded_Points` has 15 flags, one per tested concentration from lowest to highest. A `1` means the depositor removed that point before fitting.

**What goes wrong:** the flags count the 15 *tested* concentrations of that curve, not the 25 columns of the table.

```python
def points(i, use_flags=True):
    """Log concentration and response of curve i. With use_flags, the depositor's excluded points are dropped."""
    x, y = log_conc[tested[i]], Y[i, tested[i]]
    excluded = np.array(curves.at[i, "Excluded_Points"].split()) == "1"
    return (x[~excluded], y[~excluded]) if use_flags else (x, y)

n_excluded = scope["Excluded_Points"].str.count("1")
print("in-scope curves with at least one excluded point:", int((n_excluded > 0).sum()), "of", len(scope))
print("most points excluded in one curve:", int(n_excluded.max()))

example = curves[(curves["sid"] == "144203760") & (curves["replicate"] == 1)].index[0]
x_all, y_all = points(example, use_flags=False)
flags = curves.at[example, "Excluded_Points"].split()
print("\nSID 144203760, replicate 1")
print(pd.DataFrame({"conc_uM": (10 ** x_all * 1e6).round(5), "response_%": y_all, "excluded": flags}).to_string(index=False))
```

Read the example from top to bottom. The response rises to about 43 percent and then falls again at the highest concentrations. The depositor removed most of the falling part. A Hill curve can only rise to a plateau, so those points would pull the fit away.

## 8. The Hill curve

The four-parameter Hill curve has:

- `zero`: the response with no substance (bottom plateau)
- `top`: the response at very high concentration (top plateau)
- `log_ac50`: the log concentration where the response is halfway between the two
- `slope`: how steeply the curve rises

Fitting means finding the four numbers that make the curve pass as close as possible to the points (least squares).

**What goes wrong:** a fitting routine needs a starting guess. From a bad guess it can stop at a poor answer. So we work in two steps. First a coarse search: try a grid of AC50 and slope values and keep the best pair. For a fixed AC50 and slope, the best two plateaus follow from a straight-line fit, so the search is quick. Then `curve_fit` polishes that start.

We keep the AC50 inside the tested range and the slope between 0.2 and 5. The slope limits are the range of the depositor's own slopes in this record (printed below).

```python
def hill(log_c, zero, top, log_ac50, slope):
    return zero + (top - zero) / (1 + 10 ** ((log_ac50 - log_c) * slope))

fitted = curves[curves["Fit_LogAC50"].notna()]
print("depositor's Hill slopes over all", len(fitted), "fitted curves: lowest", fitted["Fit_HillSlope"].min(), "highest", fitted["Fit_HillSlope"].max())
SLOPE_MIN, SLOPE_MAX = 0.2, 5.0
SLOPE_GRID = np.array([0.2, 0.3, 0.5, 0.75, 1.0, 1.5, 2.0, 3.0, 4.0, 5.0])

def coarse_search(x, y):
    """Step 1: best (zero, top, log_ac50, slope) on a grid. Step size 0.05 in log AC50."""
    ac50_grid = np.arange(x.min(), x.max() + 1e-9, 0.05)
    A, S = np.meshgrid(ac50_grid, SLOPE_GRID, indexing="ij")
    shape = 1 / (1 + 10 ** ((A[..., None] - x) * S[..., None]))      # curve from 0 to 1 for each pair
    shape_c, y_c = shape - shape.mean(axis=-1, keepdims=True), y - y.mean()
    with np.errstate(divide="ignore", invalid="ignore"):
        span = (shape_c * y_c).sum(axis=-1) / (shape_c ** 2).sum(axis=-1)   # straight-line fit: y = zero + span * shape
    sse = ((y_c - span[..., None] * shape_c) ** 2).sum(axis=-1)
    i, j = np.unravel_index(np.nanargmin(sse), sse.shape)
    zero = y.mean() - span[i, j] * shape[i, j].mean()
    return [zero, zero + span[i, j], ac50_grid[i], SLOPE_GRID[j]]

def fit_curve(x, y):
    """Step 2: polish the coarse start with least squares. Returns zero, top, log_ac50, slope."""
    start = coarse_search(x, y)
    lower = [-np.inf, -np.inf, x.min(), SLOPE_MIN]
    upper = [np.inf, np.inf, x.max(), SLOPE_MAX]
    best, _ = curve_fit(hill, x, y, p0=start, bounds=(lower, upper))
    return best
```

## 9. Fit one curve and compare

Fit the example twice: once with the depositor's exclusions, once with all 15 points.

```python
dep = curves.loc[example]
print("deposited:        log AC50", dep["Fit_LogAC50"], "| slope", dep["Fit_HillSlope"], "| zero", dep["Fit_ZeroActivity"], "| top", dep["Fit_InfiniteActivity"])
zero, top, log_ac50, slope = fit_curve(*points(example))
print(f"refit, exclusions: log AC50 {log_ac50:.4f} | slope {slope:.3f} | zero {zero:.2f} | top {top:.2f}")
zero, top, log_ac50, slope = fit_curve(*points(example, use_flags=False))
print(f"refit, all points: log AC50 {log_ac50:.4f} | slope {slope:.3f} | zero {zero:.2f} | top {top:.2f}")
print("AC50 in uM, deposited column Potency:", dep["Potency"])
```

## 10. Fit every curve in scope

Same fit, with the depositor's exclusions, for each curve in scope.

```python
records = []
for i in scope.index:
    zero, top, log_ac50, slope = fit_curve(*points(i))
    step = np.median(np.diff(log_conc[tested[i]]))           # dilution step of this curve, in log10 units
    records.append({"check": f"s{curves.at[i, 'sid']}_r{curves.at[i, 'replicate']}", "refit_log_ac50": log_ac50, "refit_slope": slope,
                    "refit_zero": zero, "refit_top": top, "deposited_log_ac50": curves.at[i, "Fit_LogAC50"],
                    "deposited_slope": curves.at[i, "Fit_HillSlope"], "deposited_top": curves.at[i, "Fit_InfiniteActivity"], "n_excluded": int(n_excluded[i]), "step": step})
fits = pd.DataFrame(records).set_index("check")
fits["difference"] = fits["refit_log_ac50"] - fits["deposited_log_ac50"]
idx_of = dict(zip(fits.index, scope.index))                  # check id -> row of `curves`
print(len(fits), "curves fitted")
print(fits[["refit_log_ac50", "deposited_log_ac50", "difference", "refit_slope", "deposited_slope"]].head(6).round(3).to_string())
```

## 11. How close is close enough?

The deposited AC50 is the output of the depositor's fitting program. Two sound programs do not agree to the last decimal. The depositor's values also come from a limited set, as the next cell shows.

The concentration axis has a natural unit: one dilution step, the gap between neighbouring tested concentrations. This lab calls a refit a **match** when it is within **half a dilution step** of the deposited value. That places the AC50 at the same position on the titration series.

```python
print("different deposited log AC50 values among", len(fits), "curves:", scope["Fit_LogAC50"].nunique(), "| different deposited slopes:", scope["Fit_HillSlope"].nunique())
print("dilution step per curve (log10): lowest", fits["step"].min().round(4), "highest", fits["step"].max().round(4), "-> a factor of about", round(10 ** fits["step"].median(), 2))
print("half a step:", (fits["step"].median() / 2).round(3), "log units -> a factor of about", round(10 ** (fits["step"].median() / 2), 2), "in AC50")

fits["tolerance"] = [PUBLISHED[c]["tolerance"]["abs"] for c in fits.index]
fits["match"] = fits["difference"].abs() <= fits["tolerance"] + 1e-12
abs_diff = fits["difference"].abs()
print("\nmatches:", int(fits["match"].sum()), "of", len(fits))
print("median absolute difference:", abs_diff.median().round(3), "| 90th percentile:", abs_diff.quantile(0.9).round(3), "| largest:", abs_diff.max().round(3))
print("mean signed difference (refit minus deposited):", fits["difference"].mean().round(3))
```

## 12. Where the differences are

Split the curves by whether the depositor excluded points, and list the largest differences. Reading the worst cases is how you find out what the two fits do differently. Look at the slopes and at the top plateau.

```python
by_excl = fits.groupby(fits["n_excluded"] > 0).agg(curves=("match", "size"), matches=("match", "sum"), median_abs_diff=("difference", lambda d: d.abs().median()))
by_excl.index = ["no excluded points", "some excluded points"]
print(by_excl.round(3).to_string())
at_cap = fits["refit_slope"] > SLOPE_MAX - 0.01
print("\nrefit slope at the upper limit:", int(at_cap.sum()), "curves, of which matches:", int(fits.loc[at_cap, "match"].sum()))
shallow = fits["refit_slope"] < 0.6
print("refit slope below 0.6:", int(shallow.sum()), "curves, of which matches:", int(fits.loc[shallow, "match"].sum()))
top_ratio = fits["refit_top"] / fits["deposited_top"]
print("refit top plateau / deposited top plateau, median: where the AC50 matches", top_ratio[fits["match"]].median().round(2), "| where it differs", top_ratio[~fits["match"]].median().round(2))
print("\nten largest differences:")
worst = fits.reindex(abs_diff.sort_values(ascending=False).index).head(10)
print(worst[["refit_log_ac50", "deposited_log_ac50", "difference", "refit_slope", "deposited_slope", "n_excluded"]].round(3).to_string())
```

## 13. Not checked: are we fitting the same points as the depositor?

Before blaming the fit, check the inputs. The record stores an R² for the depositor's fit. Put the depositor's own four parameters through our points and compute R² again. If our points, flags and concentrations are the ones the depositor used, the two R² values agree. These numbers are for learning; the lab does not check them.

```python
def deposited_r2(i, use_flags):
    """R-squared of the depositor's own parameters on curve i, as this notebook reads the points."""
    x, y = points(i, use_flags)
    c = curves.loc[i]
    predicted = hill(x, c["Fit_ZeroActivity"], c["Fit_InfiniteActivity"], c["Fit_LogAC50"], c["Fit_HillSlope"])
    return 1 - ((y - predicted) ** 2).sum() / ((y - y.mean()) ** 2).sum()

gap_flags = np.array([abs(deposited_r2(i, True) - curves.at[i, "Fit_R2"]) for i in scope.index])
gap_all = np.array([abs(deposited_r2(i, False) - curves.at[i, "Fit_R2"]) for i in scope.index])
print("gap between recomputed and deposited R-squared, over", len(scope), "curves")
print("with the exclusion flags: median", np.median(gap_flags).round(4), "| within 0.005:", int((gap_flags < 0.005).sum()), "| more than 0.05 apart:", int((gap_flags > 0.05).sum()))
print("with all 15 points:       median", np.median(gap_all).round(4), "| within 0.005:", int((gap_all < 0.005).sum()), "| more than 0.05 apart:", int((gap_all > 0.05).sum()))
```

With the flags, the recomputed R² is within 0.05 of the deposited one for all but a few curves. With all 15 points, many more are far off. So the points, the flags and the concentration scale are read as the depositor used them, and the differences in section 12 are not caused by reading the wrong points.

## 14. Not checked: how firmly do the points fix the AC50?

For each curve, ask which AC50 values fit almost as well as the best one. Step along the log AC50 axis. At each position, find the best slope and plateaus and note the sum of squares. Keep the positions whose sum of squares is within 10 percent of the best. The width of that range says how firmly the points fix the AC50. These numbers are for learning; the lab does not check them.

```python
PROFILE_SLOPES = np.linspace(SLOPE_MIN, SLOPE_MAX, 49)

def near_best(x, y, best_sse):
    """Log AC50 grid (step 0.05) and, for each position, whether its best sum of squares is within 10 percent of best_sse."""
    ac50_grid = np.arange(x.min(), x.max() + 1e-9, 0.05)
    shape = 1 / (1 + 10 ** ((ac50_grid[:, None, None] - x) * PROFILE_SLOPES[None, :, None]))
    shape_c, y_c = shape - shape.mean(axis=-1, keepdims=True), y - y.mean()
    with np.errstate(divide="ignore", invalid="ignore"):
        span = (shape_c * y_c).sum(axis=-1) / (shape_c ** 2).sum(axis=-1)
    sse = np.nanmin(((y_c - span[..., None] * shape_c) ** 2).sum(axis=-1), axis=1)     # best over slope, per AC50
    return ac50_grid, sse <= 1.10 * min(best_sse, sse.min())

widths, deposited_inside, sharper_than_grid = [], [], []
for check, i in idx_of.items():
    x, y = points(i)
    f = fits.loc[check]
    best_sse = ((y - hill(x, f["refit_zero"], f["refit_top"], f["refit_log_ac50"], f["refit_slope"])) ** 2).sum()
    grid, ok = near_best(x, y, best_sse)
    # a very sharp optimum can fall between grid positions: then no position qualifies and the width is below the grid step
    widths.append(grid[ok].max() - grid[ok].min() if ok.any() else 0.0)
    deposited_inside.append(bool(ok[np.argmin(np.abs(grid - f["deposited_log_ac50"]))]))
    sharper_than_grid.append(not ok.any())
fits["near_best_width"], fits["deposited_near_best"] = widths, deposited_inside

print("curves whose optimum is sharper than the 0.05 grid (width counted as 0):", int(sum(sharper_than_grid)), "| of which differ from the deposit:", int((np.array(sharper_than_grid) & ~fits["match"].to_numpy()).sum()))
loose = fits["near_best_width"] > fits["step"]                 # range wider than one dilution step
differs = ~fits["match"]
print("width of the near-best range (log10), median: where the AC50 matches", fits.loc[~differs, "near_best_width"].median().round(2), "| where it differs", fits.loc[differs, "near_best_width"].median().round(2))
print("curves whose near-best range is wider than one dilution step:", int(loose.sum()), "of", len(fits), "| of which differ from the deposit:", int((loose & differs).sum()))
print("curves with a narrower range:", int((~loose).sum()), "| of which differ:", int((~loose & differs).sum()))
print("curves that differ:", int(differs.sum()), "| deposited AC50 inside the near-best range:", int(fits.loc[differs, "deposited_near_best"].sum()))
```

A curve labelled complete does not always fix its AC50. Where the near-best range is wide, very different AC50 values fit the points almost equally well, each with its own slope and top plateau, and two programs can settle on different ones. Most of the differing curves are of this kind, and they are a small share of the curves with a narrow range.

Now look at the last line. For about half of the differing curves, the deposited AC50 is itself inside the near-best range: both answers are defensible. For the other half, the deposited AC50 fits these points worse than the refit by more than 10 percent. The record does not describe how the depositor's program limits the plateaus or weights the points, so this notebook cannot say why.

## What to take away

- An AC50 is a fitted number. Before you compare two of them, find out which points each fit used, what limits it put on the slope and plateaus, and which units it reports.
- Check the inputs before the fit. Recomputing the depositor's R² from the depositor's own parameters (section 13) shows whether you are looking at the same points.
- A curve class is a label, not a guarantee. Some curves labelled complete still leave the AC50 loose (section 14). Shallow curves are the usual case: the top plateau, the slope and the AC50 can trade against each other.
- A value from a curve that never reached its top plateau is an extrapolation (section 16).
- A steep curve that rises between two neighbouring concentrations does not fix the slope. The AC50 can sit anywhere between those two concentrations.
- The concentration labels in this table are shared by all substances and are unevenly spaced near the top of the range, so a label may not be the exact concentration used for a given substance. Section 13 suggests the effect is small in this record.
- If your refit and the deposited value differ for a curve, that is a finding to look into, not a number to adjust.
